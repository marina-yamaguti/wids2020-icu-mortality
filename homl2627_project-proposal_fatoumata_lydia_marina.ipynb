## The task in our own words

When a patient arrives in an intensive care unit (ICU), the team records vital signs, lab results, the reason for admission and a few chronic conditions. Our task is to estimate the probability that the patient will **die before leaving the hospital**, using **only what is known after the first 24 hours of the ICU stay**.
The target is only known at discharge while the features are routine measurements. That is the setting where a predictive model is useful, for example to flag high-risk patients early.

## Dataset and feasibility check

**Dataset:** WiDS Datathon 2020 (Kaggle, data from MIT's GOSSIS initiative): https://www.kaggle.com/competitions/widsdatathon2020/data

### Information from preprocessing:

It has **91 713** ICU stays from 147 hospitals and **186** columns, all documented in a data dictionary. The target is binary (hospital_death = 1).

We will use **ROC-AUC** (the competition metric) to evaluate the model because only 8,6% of patients die and the accuracy (91,37%) would be misleading.

There are **missing values** in almost every row, but we noticed that patients with more tests have higher mortality, so missingness could be relevant.

To check if the task isn't too trivial, we applied AUC to the value of each feature as a prediction. Excluding APACHE, the best single feature reaches 0.74. **No feature alone solves the task.**

We found two potential sources of **leakage**:
    
- apache_4a_hospital_death_prob is the death probability from APACHE IVa, an existing clinical score. Our model could just copy it, so we remove it from the features.
- hospital_id and icu_id: mortality varies a lot between hospitals, so the model could learn each hospital’s death rate, which is useless for a new hospital. We remove these features and split the data by hospital, so no hospital appears in both the training and test sets.

## State-of-the-art and our angle

951 Kaggle teams took part in this challenge and the winners reach around 0.915 ROC-AUC with large ensembles.

We will not try to win the leaderboard. Instead, we ask whether a **model would be trustworthy in a hospital it has never seen**. That means validation grouped by hospital, a comparison with the APACHE clinical score and performance checks across subgroups (age, sex, ICU type).

We will keep the **models simple** (logistic regressioon, random forest, etc) so that every result can be explained.
